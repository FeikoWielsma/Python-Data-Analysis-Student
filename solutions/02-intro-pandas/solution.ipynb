## Step 5 : A messy file

`messy_stock_data.tsv` has a misleading extension : its fields are separated by spaces. It also
starts with three introductory lines and contains comment lines that start with `#`. Open the file
in JupyterLab's file browser before you write any code.

Read the file, whose path is in `file_messy`, into `df1` with the default options of
`pd.read_csv()` and print `df1.head()`. Read it again into `df2` with `delimiter=' '`, `header=3`
and `comment='#'`. The clean table has four rows, IBM, MSFT, GOOGLE and APPLE, and 13 columns :
`name` and the months January to December.

Write `df2` to the tab-separated file `file_clean.tsv` with `sep='\t'` and `index=False`. Read it
back with `sep='\t'` and check that the columns and values are unchanged. Then write `df2` to
`file_clean.xlsx` with `to_excel(..., index=False)`.

```python
file_messy = DATA / 'messy_stock_data.tsv'
```

```python
df1 = pd.read_csv(file_messy)
print(df1.head())

df2 = pd.read_csv(file_messy, delimiter=' ', header=3, comment='#')
print(df2)
print(df2.shape)
```

```python
df2.to_csv('file_clean.tsv', sep='\t', index=False)
restored = pd.read_csv('file_clean.tsv', sep='\t')
print(restored.equals(df2))

df2.to_excel('file_clean.xlsx', index=False)
```
