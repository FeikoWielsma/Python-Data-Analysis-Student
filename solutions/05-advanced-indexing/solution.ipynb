## Step 2 : Alignment

Read `rentals.csv` again with its month index. Uppercase all labels with
`rentals.index.str.upper()`, assign the result to `rentals.index` and print `rentals`.

Late rental reports arrive as `corrections`, a Series in a different month order. Add `corrections`
to `rentals['city_bikes']` as `aligned`. In January-to-June order the values are 322, 290, 412,
517, 622 and 680.

Drop the May correction with `corrections.drop('MAY')`, repeat the addition as `partial` and find
the month without a value. Why does the order of the corrections not matter, while a missing label
does?

````python
corrections = pd.Series({'JUN': -8, 'MAY': 12, 'APR': -5, 'MAR': 7, 'FEB': 3, 'JAN': 10})
````

````python
rentals = pd.read_csv(DATA / 'rentals.csv', index_col='month')
rentals.index = rentals.index.str.upper()
print(rentals)

aligned = rentals['city_bikes'] + corrections
print(aligned.loc[rentals.index])

partial = rentals['city_bikes'] + corrections.drop('MAY')
print(partial[partial.isna()])
````

## Step 4 : Index from a column

Read `rentals.csv` without `index_col` into `raw`. Build `rentals` with `set_index('month')` and
check `rentals.index.is_unique`. March has 405 city-bike rentals.

Reverse the rows with `raw.iloc[::-1]` and build `reordered` in the same way. Print
`reordered.loc['Mar', 'city_bikes']` and `reordered.iloc[0]['city_bikes']`. Which of the two
changed? Why would a fixed list of month names, assigned to the reversed rows, mislabel them?

Turn the index of `reordered` back into a column with `reset_index()` and print the result.

````python
raw = pd.read_csv(DATA / 'rentals.csv')
rentals = raw.set_index('month')
print(rentals.index.is_unique, rentals.loc['Mar', 'city_bikes'])

reordered = raw.iloc[::-1].set_index('month')
print(reordered.loc['Mar', 'city_bikes'], reordered.iloc[0]['city_bikes'])
print(reordered.reset_index())
````

## Step 5 : Duplicate labels

Read `rentalspercity.csv` into `by_city` with `city` as the index and check
`by_city.index.is_unique`.

Select `by_city.loc['RTM']` as `rotterdam`. How many rows does it return? Compare it with
`by_city.iloc[2]`, which selects one row by position.

Try `by_city.reindex(['AMS', 'RTM', 'UTR'])` inside `try`/`except ValueError` and print the error.
Why can't pandas pick one row per city?

````python
by_city = pd.read_csv(DATA / 'rentalspercity.csv').set_index('city')
print(by_city.index.is_unique)

rotterdam = by_city.loc['RTM']
print(rotterdam)
print(by_city.iloc[2])

try:
    by_city.reindex(['AMS', 'RTM', 'UTR'])
except ValueError as error:
    print('ValueError:', error)
````

Two items for the optional MultiIndex slides :

2. On the same index, select :
   - The Rotterdam January row as `rtm_jan` ; its city-bike value is 97

   - Every city's February row as `all_feb`, with `(slice(None), 'Feb')` and again with

````python
by_city = pd.read_csv(DATA / 'rentalspercity.csv').set_index(['city', 'month']).sort_index()
print(by_city.index.is_unique)
print(by_city)

selected = by_city.loc[['AMS', 'UTR']]
ranged = by_city.loc['AMS':'UTR']
print(len(selected), len(ranged))
````

````python
rtm_jan = by_city.loc[('RTM', 'Jan')]
print(rtm_jan['city_bikes'])

ams_utr_feb = by_city.loc[(['AMS', 'UTR'], 'Feb'), :]
print(ams_utr_feb)

all_feb = by_city.loc[(slice(None), 'Feb'), :]
print(all_feb)
print(by_city.xs('Feb', level='month')['city_bikes'].sum())
````
