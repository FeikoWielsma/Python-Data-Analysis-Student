## Step 4 : Views and copies

`original = np.arange(5, 30)` holds the integers 5 to 29. Print it in reverse order, so that the
first element becomes the last.

Take `view = original[2:5]` and an independent copy of the same slice with `.copy()`. Set
`view[0] = -1` and print the first five values of `original` and the copy. Check with
`np.shares_memory()` which of the two shares memory with `original`. When do you need a copy before
you change a slice?

```python
original = np.arange(5, 30)
```

```python
print(original[::-1])

view = original[2:5]
independent = view.copy()
view[0] = -1
print(original[:5], independent)
print(np.shares_memory(original, view), np.shares_memory(original, independent))
```

## Step 10 : Z-scores

`scores = np.array([[70, 80], [90, 60], [80, 70]])` holds two test scores, in the columns, for three
students, in the rows. Standardize each column : subtract the column mean and divide by the column
standard deviation, both with `axis=0`. Which student, by row position, lies furthest above the
average in the first test? Use `argmax`.

```python
scores = np.array([[70, 80], [90, 60], [80, 70]])
```

```python
z_scores = (scores - scores.mean(axis=0)) / scores.std(axis=0)
print(z_scores.round(2))
print(z_scores[:, 0].argmax())
```
