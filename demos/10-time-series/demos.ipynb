## demo04 · Date range and gaps

```python
import pandas as pd

weather = pd.read_csv(
    "weather_data_austin_2010.csv",
    parse_dates=["Date"],
    date_format="%Y%m%d %H:%M",
    index_col="Date",
)
evening = pd.date_range("2026-02-11 20:00", periods=4, freq="h")
print(evening)

expected = pd.date_range(weather.index.min(), weather.index.max(), freq="h")
missing = expected.difference(weather.index)
print(len(expected), len(weather))
print(missing)
```

## demo05 · Reindex and fill

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"], index_col="Date")
evening = pd.date_range("2026-02-11 20:00", periods=4, freq="h")
print(sales.reindex(evening))
print(sales.reindex(evening, method="ffill"))
print(sales.reindex(evening, method="bfill"))
```

## demo07 · Check a resampled value

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"], index_col="Date")
daily_mean = sales["Units"].resample("D").mean()
print(daily_mean.loc["2026-02-02"])
print(sales.loc["2026-02-02", "Units"])
print(sales.loc["2026-02-02", "Units"].mean())
```

## demo08 · Resample sum

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"], index_col="Date")
daily_sum = sales["Units"].resample("D").sum()
print(daily_sum.head(8))
print(sales["Units"].resample("D").sum().max())
print(sales["Units"].resample("D").sum().idxmax())
```

## demo09 · Empty intervals

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"], index_col="Date")
daily = sales["Units"].resample("D")
print(daily.agg(["count", "sum", "mean"]).loc["2026-02-05":"2026-02-08"])
print(sales["Units"].resample("D").sum(min_count=1).loc["2026-02-05":"2026-02-08"])
```

## demo10 · Resample week

## demo11 · Binning events

## demo12 · Upsampling

```python
import pandas as pd

weather = pd.read_csv(
    "weather_data_austin_2010.csv",
    parse_dates=["Date"],
    date_format="%Y%m%d %H:%M",
    index_col="Date",
)
daily = weather["Temperature"].resample("D").mean().head(2).round(1)
print(daily)
print(daily.resample("6h").asfreq())
print(daily.resample("6h").ffill())
print(daily.resample("6h").interpolate())
```

## demo13 · Clock gap

```python
import pandas as pd

weather = pd.read_csv(
    "weather_data_austin_2010.csv",
    parse_dates=["Date"],
    date_format="%Y%m%d %H:%M",
    index_col="Date",
)
hourly = weather["Temperature"].asfreq("h")
print(hourly.isna().sum())
print(hourly.loc["2010-03-14 01:00":"2010-03-14 05:00"])
```

## demo14 · Gap filling

```python
import pandas as pd

weather = pd.read_csv(
    "weather_data_austin_2010.csv",
    parse_dates=["Date"],
    date_format="%Y%m%d %H:%M",
    index_col="Date",
)
hourly = weather["Temperature"].asfreq("h")
print(hourly.ffill().loc["2010-03-14 02:00":"2010-03-14 04:00"])
print(hourly.interpolate(method="time").loc["2010-03-14 02:00":"2010-03-14 04:00"])
```

## demo15 · Rolling

```python
import pandas as pd

aapl = pd.read_csv("AAPL.csv", parse_dates=["Date"], index_col="Date")
close = aapl["Close"].tail(8)
table = pd.DataFrame(
    {
        "close": close,
        "rolling": close.rolling(window=4).mean(),
        "partial": close.rolling(window=4, min_periods=2).mean(),
        "centered": close.rolling(window=3, center=True).mean(),
    }
)
print(table.round(2))
```

## demo16 · shift, diff, pct_change

```python
import pandas as pd

aapl = pd.read_csv("AAPL.csv", parse_dates=["Date"], index_col="Date")
close = aapl["Close"].tail(5)
print(
    pd.DataFrame(
        {
            "close": close,
            "previous": close.shift(1),
            "change": close.diff(),
            "pct": close.pct_change().mul(100).round(2),
        }
    )
)
```

## demo17 · Plot a time series

```python
import matplotlib.pyplot as plt
import pandas as pd

aapl = pd.read_csv("AAPL.csv", parse_dates=["Date"], index_col="Date")
close = aapl["Close"]

fig, ax = plt.subplots(figsize=(8, 3.5))
close.plot(ax=ax, alpha=0.5, label="weekly close")
close.rolling(window=13).mean().plot(ax=ax, label="13-week rolling mean")
close.resample("QS").mean().plot(ax=ax, drawstyle="steps-post", label="quarterly mean")
ax.set(title="Apple share price", xlabel="", ylabel="US dollars")
ax.legend()
fig.tight_layout()
fig.savefig("price-rolling.png")
plt.show()
```

## demo18 · dt accessor

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"])
print(sales.head(3))
print(sales["Date"].dt.hour.head(3).tolist())
print(sales["Date"].dt.day_name().head(3).tolist())
print(sales["Date"].dt.strftime("%d %b %H:%M").head(3).tolist())
print(
    sales.groupby(sales["Date"].dt.day_name())["Units"].sum().sort_values(ascending=False).head(3)
)
```

## demo19 · Boolean arithmetic

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"])
is_ware = sales["Product"].str.contains("ware")
print(True + True, True + False)
print(is_ware.sum(), len(sales))
print(is_ware.mean().round(3))
```

## demo20 · Time zones

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"])
central = sales["Date"].dt.tz_localize("America/Chicago")
print(central.head(3))
eastern = central.dt.tz_convert("America/New_York")
print(eastern.head(3))
print(sales["Date"].iloc[7], "->", eastern.iloc[7])
```

## demo21 · UTC

```python
import pandas as pd

sales = pd.read_csv("sales-feb-2026.csv", parse_dates=["Date"], index_col="Date")
print(sales.index.is_monotonic_increasing, sales.index.has_duplicates)
utc = sales.index.tz_localize("America/Chicago").tz_convert("UTC")
print(utc[:3])
```

## demo22 · DST transitions

```python
import pandas as pd

times = pd.Series(pd.to_datetime(["2026-03-08 02:30", "2026-11-01 01:30"]))
print(times.dt.tz_localize("America/Chicago", nonexistent="shift_forward", ambiguous="NaT"))
try:
    times.dt.tz_localize("America/Chicago")
except Exception as error:
    print(type(error).__name__, error)
```
