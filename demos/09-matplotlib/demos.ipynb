## demo01 · Two interfaces

```python
import matplotlib.pyplot as plt
import pandas as pd

degrees = pd.read_csv("women-degrees.csv")

# pyplot interface: functions act on the current figure
plt.plot(degrees["year"], degrees["computer_science"])
plt.title("Computer Science")
plt.show()

# Object-oriented interface: methods on explicit Figure and Axes objects
fig, ax = plt.subplots()
ax.plot(degrees["year"], degrees["computer_science"])
ax.set_title("Computer Science")
fig.savefig("two-interfaces.png")
plt.show()
```

## demo03 · Line styles

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2, 20)
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, x, color="tab:blue", linestyle="-", label="linear")
ax.plot(x, x**2, color="tab:orange", linestyle="--", marker="o", label="quadratic")
ax.plot(x, x**3, color="tab:green", linestyle=":", linewidth=3, label="cubic")
ax.plot(x, 2 * x, "k^-", markersize=4, label="format string 'k^-'")
ax.legend()
fig.tight_layout()
fig.savefig("line-styles.png")
plt.show()
```

## demo04 · Legend

## demo05 · Subplots

```python
import matplotlib.pyplot as plt
import pandas as pd

degrees = pd.read_csv("women-degrees.csv")

fields = ["physical_sciences", "computer_science", "health", "education"]
fig, axes = plt.subplots(2, 2, figsize=(7, 5), sharex=True, layout="constrained")
for ax, field in zip(axes.flat, fields):
    ax.plot(degrees["year"], degrees[field])
    ax.set_title(field.replace("_", " ").title())
axes[0, 0].set_ylabel("% women")
axes[1, 0].set_ylabel("% women")
fig.savefig("subplots.png")
plt.show()
```

## demo06 · Limits and scales

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

degrees = pd.read_csv("women-degrees.csv")


fig, (left, right) = plt.subplots(1, 2, figsize=(8, 3.5), layout="constrained")
left.plot(degrees["year"], degrees["computer_science"])
left.set_xlim(1990, 2010)
left.set_ylim(0, 50)
left.set_xticks([1990, 2000, 2010])
left.set_title("Limits and ticks")

x = np.arange(1, 9)
right.plot(x, 10.0**x, marker="o")
right.set_yscale("log")
right.set_title("Log scale")
fig.savefig("limits-and-scales.png")
plt.show()
```

## demo07 · Annotate

```python
import matplotlib.pyplot as plt
import pandas as pd

degrees = pd.read_csv("women-degrees.csv")

cs = degrees["computer_science"]
peak_year = degrees.loc[cs.idxmax(), "year"]

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(degrees["year"], cs)
ax.annotate(
    f"Peak: {cs.max()}% in {peak_year}",
    xy=(peak_year, cs.max()),
    xytext=(peak_year + 5, cs.max() + 3),
    arrowprops={"arrowstyle": "->", "color": "black"},
)
ax.text(1971, 15, "Source: US degree statistics", fontsize=9, color="gray")
ax.set(ylim=(0, 45), xlabel="Year", ylabel="% of degrees awarded to women")
fig.tight_layout()
fig.savefig("annotate.png")
plt.show()
```

## demo08 · Scatter

```python
import matplotlib.pyplot as plt
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")

fig, ax = plt.subplots(figsize=(6, 4))
points = ax.scatter(auto["weight"], auto["mpg"], c=auto["hp"], cmap="viridis", s=20, alpha=0.8)
fig.colorbar(points, ax=ax, label="Horsepower")
ax.set(title="Heavier cars use more fuel", xlabel="Weight (lb)", ylabel="Miles per gallon")
fig.tight_layout()
fig.savefig("scatter.png")
plt.show()
```

## demo09 · Histogram

```python
import matplotlib.pyplot as plt
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(auto["mpg"], bins=20, edgecolor="white")
ax.axvline(auto["mpg"].median(), color="black", linestyle="--", label="median")
ax.set(title="Fuel economy of 392 cars", xlabel="Miles per gallon", ylabel="Number of cars")
ax.legend()
fig.tight_layout()
fig.savefig("histogram.png")
plt.show()
```

## demo10 · Box plot

```python
import matplotlib.pyplot as plt
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")

origins = ["US", "Europe", "Asia"]
groups = [auto.loc[auto["origin"] == origin, "mpg"] for origin in origins]

fig, ax = plt.subplots(figsize=(6, 4))
ax.boxplot(groups, tick_labels=origins)
ax.set(title="Fuel economy by origin", ylabel="Miles per gallon")
fig.tight_layout()
fig.savefig("boxplot.png")
plt.show()
```

## demo11 · Bar

```python
import matplotlib.pyplot as plt
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")

mean_mpg = auto.groupby("origin")["mpg"].mean().sort_values()
counts = auto["cyl"].value_counts().sort_index()

fig, (left, right) = plt.subplots(1, 2, figsize=(8, 3.5), layout="constrained")
left.barh(mean_mpg.index, mean_mpg.to_numpy())
left.set(title="Mean mpg by origin", xlabel="Miles per gallon")
right.bar(counts.index.astype(str), counts.to_numpy())
right.set(title="Cars by cylinders", xlabel="Cylinders", ylabel="Cars")
fig.savefig("bar.png")
plt.show()
```

## demo12 · Pie

```python
import matplotlib.pyplot as plt
import pandas as pd

titanic = pd.read_csv("titanic.csv")
counts = titanic["pclass"].value_counts().sort_index()

fig, ax = plt.subplots(figsize=(4.5, 4.5))
ax.pie(
    counts.to_numpy(),
    labels=["1st class", "2nd class", "3rd class"],
    autopct="%1.0f%%",
    startangle=90,
    counterclock=False,
)
ax.set_title("Titanic passengers by class")
fig.savefig("pie.png")
plt.show()
```

## demo13 · Dates

```python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import pandas as pd

aapl = pd.read_csv("AAPL.csv", parse_dates=["Date"])

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.plot(aapl["Date"], aapl["Close"])
locator = mdates.AutoDateLocator()
ax.xaxis.set_major_locator(locator)
ax.xaxis.set_major_formatter(mdates.ConciseDateFormatter(locator))
ax.set(title="Apple weekly close", ylabel="US dollars")
fig.tight_layout()
fig.savefig("dates.png")
plt.show()
```

## demo14 · Contour

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 200)
y = np.linspace(-3, 3, 200)
X, Y = np.meshgrid(x, y)
Z = np.exp(-(X**2 + Y**2) / 2) - 0.6 * np.exp(-((X - 1.5) ** 2 + (Y - 1) ** 2))

fig, ax = plt.subplots(figsize=(6, 4.5))
filled = ax.contourf(X, Y, Z, levels=15, cmap="viridis")
lines = ax.contour(X, Y, Z, levels=8, colors="black", linewidths=0.6)
ax.clabel(lines, fontsize=8)
fig.colorbar(filled, ax=ax, label="height")
ax.set(title="Contour plot", xlabel="x", ylabel="y")
fig.tight_layout()
fig.savefig("contour.png")
plt.show()
```

## demo15 · Surface 3D

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 80)
y = np.linspace(-3, 3, 80)
X, Y = np.meshgrid(x, y)
Z = np.exp(-(X**2 + Y**2) / 2) - 0.6 * np.exp(-((X - 1.5) ** 2 + (Y - 1) ** 2))

fig, ax = plt.subplots(figsize=(6, 4.5), subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, cmap="viridis", edgecolor="none")
ax.set(title="Surface plot", xlabel="x", ylabel="y", zlabel="height")
fig.tight_layout()
fig.savefig("surface-3d.png")
plt.show()
```

## demo16 · Styles

## demo23 · Seaborn levels

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

tips = pd.read_csv("tips.csv")

fig, ax = plt.subplots()
sns.boxplot(data=tips, x="day", y="tip", ax=ax)

g = sns.catplot(data=tips, x="day", y="tip", col="time", kind="box", height=3.5)
print(type(g).__name__)
g.savefig("seaborn-catplot.png")
plt.show()
```

## demo24 · Seaborn distributions

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

tips = pd.read_csv("tips.csv")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5), layout="constrained")
sns.histplot(data=tips, x="total_bill", hue="time", stat="count", ax=axes[0])
sns.histplot(data=tips, x="total_bill", hue="time", stat="density", common_norm=False, ax=axes[1])
sns.kdeplot(data=tips, x="total_bill", hue="time", common_norm=False, ax=axes[1])
axes[0].set_title('stat="count"')
axes[1].set_title('stat="density", common_norm=False')
fig.savefig("seaborn-distributions.png")
plt.show()
```

## demo25 · Seaborn estimates

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

titanic = pd.read_csv("titanic.csv")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5), layout="constrained")
sns.barplot(
    data=titanic,
    x="pclass",
    y="survived",
    hue="sex",
    estimator="mean",
    errorbar=("ci", 95),
    seed=0,
    ax=axes[0],
)
sns.countplot(data=titanic, x="pclass", hue="sex", ax=axes[1])
axes[0].set(title="Survival rate, 95% CI", ylabel="share survived")
axes[1].set(title="Passengers", ylabel="count")
fig.savefig("seaborn-estimates.png")

rates = titanic.groupby(["pclass", "sex"])["survived"].agg(["size", "mean"])
print(rates.round(3))
plt.show()
```

## demo27 · Seaborn facets

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

tips = pd.read_csv("tips.csv")

sns.set_theme(style="whitegrid")
g = sns.catplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex",
    col="time",
    kind="bar",
    errorbar=("ci", 95),
    seed=0,
    height=3.5,
)
g.savefig("seaborn-facets.png")

print(tips.groupby(["time", "day", "sex"], observed=True).size())
plt.show()
```
