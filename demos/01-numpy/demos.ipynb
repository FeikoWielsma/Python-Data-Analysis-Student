## demo01 · numpy_intro

```python
import numpy as np

# We have a list with values, e.g. temperatures in Celsius:
cvalues = [25.3, 24.8, 26.9, 23.9]

# We will turn this into a one-dimensional 01-numpy array:
C = np.array(cvalues)
print(C)

# Let's assume, we want to turn the values into degrees Fahrenheit.
# This is very easy to accomplish with a 01-numpy array.
# The solution to our problem can be achieved by simple scalar multiplication:
print(C * 9 / 5 + 32)

# Compared to this, the solution for our Python list is extremely awkward:
fvalues = [x * 9 / 5 + 32 for x in cvalues]
print(fvalues)
```

## demo02 · Time Comparison

```python
import timeit

import numpy as np

# One of the main advantages of NumPy is its advantage in time compared to standard Python.
# Both versions below add the same two sequences element by element.
# The inputs are created once, outside the timed code, so only the addition is timed.
# Each version is timed several times and the fastest run is reported.

size_of_vec = 1_000_000
REPEAT = 5

X_list = list(range(size_of_vec))
Y_list = list(range(size_of_vec))
X_array = np.arange(size_of_vec)
Y_array = np.arange(size_of_vec)


def pure_python_version():
    Z = []
    for i in range(len(X_list)):
        Z.append(X_list[i] + Y_list[i])
    return Z


def numpy_version():
    return X_array + Y_array


def seconds_per_call(func):
    timer = timeit.Timer(func)
    loops, _ = timer.autorange()  # enough calls for a measurable total
    return min(timer.repeat(repeat=REPEAT, number=loops)) / loops


# Compare like with like: both versions must give the same answer.
print("Same result:", np.array_equal(np.array(pure_python_version()), numpy_version()))

t1 = seconds_per_call(pure_python_version)
t2 = seconds_per_call(numpy_version)
print(f"Python loop: {t1 * 1000:8.3f} ms per call (best of {REPEAT})")
print(f"NumPy      : {t2 * 1000:8.3f} ms per call (best of {REPEAT})")
print(f"On this machine NumPy was about {t1 / t2:.0f} times faster for {size_of_vec:,} elements.")
```

## demo04 · numpyversusarrayperformance.

```python
import timeit

import numpy as np

# Add two sequences of SIZE numbers: a list comprehension versus a NumPy array.
# Inputs are built once before timing; only the addition (including creating
# the result list or array) is timed. Times are the best of REPEAT runs.

SIZE = 1_000_000
REPEAT = 5

L1 = list(range(SIZE))
L2 = list(range(SIZE))
A1 = np.arange(SIZE)
A2 = np.arange(SIZE)


def list_version():
    return [x + y for x, y in zip(L1, L2)]


def array_version():
    return A1 + A2


def ms_per_call(func):
    timer = timeit.Timer(func)
    loops, _ = timer.autorange()
    return min(timer.repeat(repeat=REPEAT, number=loops)) / loops * 1000


print("Same result:", np.array_equal(np.array(list_version()), array_version()))

list_ms = ms_per_call(list_version)
array_ms = ms_per_call(array_version)
print(f"List comprehension: {list_ms:8.3f} ms")
print(f"NumPy addition    : {array_ms:8.3f} ms")
print(f"On this machine the NumPy version was about {list_ms / array_ms:.0f} times faster.")
```

## demo12 · Iteration Order

```python
# Controlling Iteration Order:

# There are times when it is important to visit the elements of an array in a
# specific order, irrespective of the layout of the elements in memory. The nditer
# object provides an order parameter to control this aspect of iteration.
# The default, having the behavior described above, is order='K' to keep the existing order.
# This can be overridden with order='C' for C order and order='F' for Fortran order.

# Python program for iterating over array # using particular order

import numpy as np

# creating an array using arrange method
a = np.arange(12)

# shape array with 3 rows and 4 columns
a = a.reshape(3, 4)

print("Original array is:")
print(a)


print("Iteration an array in C-style order:")
for x in np.nditer(a, order="C"):
    print(x, end=" ")

print()

print("Iteration an array in F-style order:")
for x in np.nditer(a, order="F"):
    print(x, end=" ")
```

## demo13 · CandF Order Performance

```python
import argparse
import timeit

import numpy as np

# Sum the same values along each axis, stored in C (row-major) and F (column-major) order.
# Default size is 2000 x 2000 (two arrays, about 64 MB in total).
# Use --large for the original 10000 x 10000 (two arrays, about 1.6 GB in total).

parser = argparse.ArgumentParser(description="C versus F order: sum along each axis")
parser.add_argument("--large", action="store_true", help="use the original 10000 x 10000 size")
parser.add_argument("--size", type=int, default=2000, help="rows and columns (default 2000)")
args = parser.parse_args([])
n = 10_000 if args.large else args.size
REPEAT = 5

rng = np.random.default_rng(42)
c = rng.standard_normal((n, n))  # C order is the default
f = np.asfortranarray(c)  # same values, F order
print(c.shape, f"{(c.nbytes + f.nbytes) / 1e6:.0f} MB in total")

for axis in (0, 1):
    # Same values, same result; only the memory layout differs
    print("Same sums:", np.allclose(c.sum(axis=axis), f.sum(axis=axis)))


def ms_per_call(func):
    timer = timeit.Timer(func)
    loops, _ = timer.autorange()
    return min(timer.repeat(repeat=REPEAT, number=loops)) / loops * 1000


for name, a in (("C", c), ("F", f)):
    for axis in (0, 1):
        ms = ms_per_call(lambda: a.sum(axis=axis))
        print(f"{name} order axis = {axis}: {ms:8.3f} ms per sum")

print("Timings are observations on this machine; compare the four lines with each other.")
```

## demo19 · Ravel and Stack

```python
import numpy as np

x = np.array([(1, 2, 3), (3, 4, 5)])
print(x.ravel())  # [1 2 3 3 4 5]

x = np.array([(1, 2, 3), (3, 4, 5)])
y = np.array([(1, 2, 3), (3, 4, 5)])

print(np.vstack((x, y)))
# [[1 2 3]
#  [3 4 5]
#  [1 2 3]
#  [3 4 5]]

print(np.hstack((x, y)))
# [[1 2 3 1 2 3]
#  [3 4 5 3 4 5]]

r = x.ravel()  # a view when possible
f = x.flatten()  # always a copy
print(np.shares_memory(x, r), np.shares_memory(x, f))
```
