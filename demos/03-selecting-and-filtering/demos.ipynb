```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.loc["May", "spam"])
print(df.loc["Jan", "salt"])
```

## demo03 · loc slices

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.loc["Jan":"Apr", :])
print(df.loc[:, "eggs":"salt"])
```

## demo04 · loc lists

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.loc[["Jan", "Jun"], ["eggs", "spam"]])
print(df.loc["Feb":"Apr", ["spam", "eggs"]])
```

## demo05 · iloc

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.iloc[4, 2])
print(df.iloc[2:5, 1:])
print(df.loc["Mar":"May", "salt":"spam"])
```

## demo06 · iloc lists

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.iloc[[0, 4, 5], [0, 2]])
print(df.iloc[-2:, :])
```

## demo07 · Selecting columns

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df["eggs"])
print(df[["salt", "eggs"]])
```

## demo08 · Series versus DataFrame

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(type(df["eggs"]).__name__, df["eggs"].shape)
print(type(df[["eggs"]]).__name__, df[["eggs"]].shape)
```

## demo09 · Selecting from a Series

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
eggs = df["eggs"]
print(eggs.loc["Feb":"Apr"])
print(eggs.iloc[0], eggs.iloc[-1])
```

## demo10 · Shortcuts

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df["salt"]["Jan"])
print(df.eggs["Mar"])
print(df.loc["Jan", "salt"], df.loc["Mar", "eggs"])
```

## demo11 · Boolean Series

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
mask = df["salt"] > 60
print(mask)
```

## demo12 · Filtering

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df[df["salt"] > 60])
enough_salt = df["salt"] > 60
print(df.loc[enough_salt, ["salt", "spam"]])
```

## demo13 · Combining filters

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df[(df["salt"] >= 50) & (df["eggs"] < 200)])
print(df[(df["salt"] >= 50) | (df["eggs"] < 200)])
print(df[~(df["salt"] >= 50)])
```

## demo14 · isin and between

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df[df["eggs"].between(100, 210)])
first_quarter = df.index.isin(["Jan", "Feb", "Mar"])
print(df[first_quarter])
```

## demo16 · Sorting

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop.sort_values("units", ascending=False).head(3))
print(shop.sort_values(["product", "units"]).head(4))
```

## demo17 · Largest

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop.nlargest(3, "units"))
print(shop["units"].idxmax(), shop["units"].max())
print(shop.loc[shop["units"].idxmax(), "order_id"])
```

## demo18 · Value counts

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop["product"].value_counts())
print(shop["product"].value_counts(normalize=True).round(2))
print(shop["city"].value_counts())
```

## demo19 · unique

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop["product"].unique().tolist())
print(shop["product"].nunique())
print(shop["units"].median(), shop["units"].mean())
```
