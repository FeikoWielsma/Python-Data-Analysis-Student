## demo01 · Labels from data

```python
import pandas as pd

sales = pd.read_csv("sales.csv")
print(sales.head(3))
sales = sales.set_index("month")
print(sales.index.is_unique)
print(sales.loc["Feb", "eggs"], sales.iloc[1, 0])
```

## demo02 · Labels and positions

```python
import pandas as pd

prices = pd.Series([10.70, 10.86, 10.74], index=["Mon", "Tue", "Wed"])
print(prices.loc["Tue"], prices.iloc[1])
print(prices.loc["Mon":"Tue"].tolist())
print(prices.iloc[0:2].tolist())

numbered = pd.Series([10.70, 10.86, 10.74], index=[3, 2, 1])
print(numbered.loc[1], numbered.iloc[1])
```

## demo03 · Replacing labels

```python
import pandas as pd

shares = pd.Series([10.70, 10.86], index=["Mon", "Tue"])
try:
    shares.index[0] = "Monday"
except TypeError as error:
    print("TypeError:", error)

shares = shares.rename(index={"Mon": "Monday", "Tue": "Tuesday"})
print(shares)

shares.index = ["M", "T"]
print(shares.index.tolist())
```

## demo04 · Transforming labels

```python
import pandas as pd

sales = pd.read_csv("sales.csv", index_col="month")
print(sales.index)
sales.index = sales.index.str.upper()
print(sales.index)
sales.index = sales.index.map(lambda month: month[0])
print(sales.index.tolist())
```

## demo05 · Index names

```python
import pandas as pd

sales = pd.read_csv("sales.csv", index_col="month")
sales.index.name = "MONTHS"
sales.columns.name = "PRODUCTS"
print(sales.head(3))
print(sales.reset_index().columns.tolist())
```

## demo06 · Alignment

```python
import pandas as pd

sales = pd.Series([47, 110], index=["Jan", "Feb"], name="eggs")
adjustment = pd.Series([2, 1], index=["Feb", "Jan"])
print(sales + adjustment)

partial = pd.Series([2], index=["Feb"])
print(sales + partial)
print(sales.add(partial, fill_value=0))
```

## demo07 · reindex

```python
import pandas as pd

sales = pd.read_csv("sales.csv", index_col="month")
print(sales.reindex(["Jun", "Jan", "Jul"]))
print(sales["eggs"].reindex(["Jan", "Jul"], fill_value=0))
```

## demo08 · Duplicate labels

```python
import pandas as pd

sales = pd.DataFrame(
    {"state": ["CA", "NY", "NY"], "month": ["Jan", "Jan", "Feb"], "eggs": [47, 221, 77]}
).set_index("state")
print(sales.index.is_unique)
print(sales.loc["NY"])
print(sales.iloc[1])
try:
    sales.reindex(["CA", "NY"])
except ValueError as error:
    print("ValueError:", error)
```

## demo09 · MultiIndex

```python
import pandas as pd

rows = pd.DataFrame(
    {"state": ["CA", "CA", "NY"], "month": ["Jan", "Feb", "Jan"], "eggs": [47, 110, 221]}
)
by_state_month = rows.set_index(["state", "month"])
print(by_state_month)
print(by_state_month.index.is_unique, by_state_month.index.names)
print(by_state_month.loc[("CA", "Feb"), "eggs"])
```

## demo10 · Set and sort MultiIndex

```python
import pandas as pd

stocks = pd.read_csv("stocks.csv")
print(stocks)
stocks = stocks.set_index(["Symbol", "Date"]).sort_index()
print(stocks)
```

## demo11 · Outer level

```python
import pandas as pd

stocks = pd.read_csv("stocks.csv").set_index(["Symbol", "Date"]).sort_index()
print(stocks.loc["AAPL"])
print(stocks.loc[("CSCO", "2026-10-06"), "Close"])
print(stocks.loc["CSCO":"MSFT", "Close"])
```

## demo12 · Lists within levels

```python
import pandas as pd

stocks = pd.read_csv("stocks.csv").set_index(["Symbol", "Date"]).sort_index()
print(stocks.loc[(["AAPL", "MSFT"], "2026-10-07"), :])
print(stocks.loc[("CSCO", ["2026-10-07", "2026-10-05"]), "Close"])
```

## demo13 · Inner level

```python
import pandas as pd

stocks = pd.read_csv("stocks.csv").set_index(["Symbol", "Date"]).sort_index()
print(stocks.loc[(slice(None), "2026-10-06"), :])
print(stocks.xs("2026-10-06", level="Date"))
```
