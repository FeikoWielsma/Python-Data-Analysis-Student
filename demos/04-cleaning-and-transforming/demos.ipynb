## demo01 · Zeros and NaNs

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
df2 = df.copy()
df2["bacon"] = [0, 0, 50, 60, 70, 80]
print(df2)
```

## demo02 · all and any

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
df2 = df.copy()
df2["bacon"] = [0, 0, 50, 60, 70, 80]
print(df2.all())
print(df2.loc[:, df2.all()])
```

## demo03 · Finding missing values

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
df2 = df.copy()
df2["bacon"] = [0, 0, 50, 60, 70, 80]
print(df2.isna().sum())
print(df2.loc[:, df2.isna().any()])
print(df2[df2["salt"].isna()])
```

## demo04 · Dropping missing values

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.dropna())
print(df.dropna(subset=["eggs", "spam"]).shape)
print(df.dropna(axis="columns").columns.tolist())
print(df.dropna(axis="columns", thresh=5).columns.tolist())
```

## demo05 · Filling missing values

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.fillna(0).loc["May"])
print(df["salt"].fillna(df["salt"].median()).loc["May"])
```

## demo06 · Missing values and dtypes

```python
import pandas as pd

visits = pd.DataFrame({"city": ["Paris", "Berlin", "Rome"], "visitors": [139, 237, 326]})
print(visits["visitors"].dtype)
visits.loc[1, "visitors"] = float("nan")
print(visits["visitors"].dtype)
print(visits)

counts = pd.Series([139, None, 326], dtype="Int64")
print(counts)
```

## demo07 · Text and nullable dtypes

```python
import pandas as pd

text = pd.Series(["Paris", None])
names = pd.Series(["Paris", None], dtype="string")
counts = pd.Series([7, None], dtype="Int64")
flags = pd.Series([True, None], dtype="boolean")
print(text.dtype, names.dtype, counts.dtype, flags.dtype)
print(text.isna().tolist(), counts.isna().tolist())
print((counts > 5).tolist())
```

## demo08 · Modifying with loc

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
df2 = df.copy()
mask = df2["salt"] > 55
print(df2.loc[mask, "eggs"])
df2.loc[mask, "eggs"] += 5
print(df2)
```

## demo09 · Chained assignment

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
df2 = df.copy()
mask = df2["salt"] > 55
df2["eggs"][mask] += 5
print(df2["eggs"].tolist())
```

## demo10 · Copy-on-Write

```python
import pandas as pd

df = pd.DataFrame({"visitors": [139, 237]})
column = df["visitors"]
column.iloc[0] = 0
print(df["visitors"].tolist())
df.loc[0, "visitors"] = 0
print(df["visitors"].tolist())
```

## demo11 · Vectorized arithmetic

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df // 12)
print(df["eggs"] + df["spam"])
print(df["eggs"].add(df["salt"], fill_value=0))
```

## demo12 · NumPy functions

```python
import numpy as np
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")

print(np.sqrt(df["eggs"]).round(2))
print(np.log10(df).round(2))
```

## demo13 · apply

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")


def dozens(n):
    return n // 12


print(df.apply(dozens))
print(df.apply(lambda column: type(column).__name__))
```

## demo14 · Row apply and map

```python
import pandas as pd

df = pd.read_csv("sales.csv", index_col="month")
print(df.apply(lambda row: row.max() - row.min(), axis=1))
print(df.map(lambda value: f"{value:.0f}").loc["Jan"])
```

## demo15 · map with a dictionary

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
kind = {"Coffee": "drink", "Tea": "drink"}
shop["kind"] = shop["product"].map(kind)
print(shop[["product", "kind"]].head(4))
```

## demo16 · astype

```python
import pandas as pd

raw = pd.DataFrame({"units": ["3", "1", "2"], "price": ["4.50", "3.20", "n/a"]})
print(raw.dtypes)
raw["units"] = raw["units"].astype("int64")
raw["price"] = pd.to_numeric(raw["price"], errors="coerce")
print(raw.dtypes)
print(raw)
```

## demo17 · Rename and drop

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
orders = shop.rename(columns={"units": "quantity"}).drop(columns=["date"])
print(orders.columns.tolist())
print(shop.columns.tolist())
```

## demo18 · Cleaning text

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop["city"].unique().tolist())
shop["city"] = shop["city"].str.strip().str.title()
print(shop["city"].value_counts())
```

## demo19 · Text tests

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
print(shop["product"].str.startswith("C").tolist())
print(shop.loc[shop["product"].str.contains("a"), "product"].unique().tolist())
print(shop["order_id"].str.fullmatch(r"A\d{3}").all())
print(shop["order_id"].str[1:].str.isdigit().all())
print(shop["order_id"].str.lower().head(2).tolist())
```

## demo20 · Storing a transformation

```python
import pandas as pd

shop = pd.read_csv("shop.csv")
shop["city"] = shop["city"].str.strip().str.title()
shop["revenue"] = shop["units"] * shop["price"]
print(shop[["order_id", "city", "units", "price", "revenue"]].head(4))
print(shop["revenue"].sum().round(2))
```
