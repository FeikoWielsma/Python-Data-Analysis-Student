## demo01 · Series

## demo02 · From a dictionary

```python
import pandas as pd

data = {
    "weekday": ["Sun", "Sun", "Mon", "Mon"],
    "city": ["Paris", "Berlin", "Paris", "Berlin"],
    "visitors": [139, 237, 326, 456],
    "signups": [7, 12, 3, 5],
}
users = pd.DataFrame(data)
print(users)
```

## demo03 · From lists

```python
import pandas as pd

cities = ["Paris", "Berlin", "Paris", "Berlin"]
signups = [7, 12, 3, 5]
visitors = [139, 237, 326, 456]
weekdays = ["Sun", "Sun", "Mon", "Mon"]
labels = ["city", "signups", "visitors", "weekday"]
columns = [cities, signups, visitors, weekdays]
users = pd.DataFrame(dict(zip(labels, columns)))
print(users)
```

## demo05 · Broadcasting dictionary

```python
import pandas as pd

heights = [59.0, 65.2, 62.9, 65.4, 63.7, 65.7, 64.1]
results = pd.DataFrame({"height": heights, "sex": "M"})
print(results)
```

## demo06 · Change index and columns

```python
import pandas as pd

heights = [59.0, 65.2, 62.9, 65.4, 63.7, 65.7, 64.1]
results = pd.DataFrame({"height": heights, "sex": "M"})
results.columns = ["height (in)", "sex"]
results.index = list("ABCDEFG")
print(results.head(3))

results = results.rename(columns={"height (in)": "height_in"})
print(results.columns.tolist())
```

## demo18 · Assemble dates

```python
import pandas as pd

filepath = "ISSN_D_tot.csv"
col_names = ["year", "month", "day", "dec_date", "sunspots", "definite"]
sunspots = pd.read_csv(filepath, header=None, names=col_names, na_values={"sunspots": [" -1"]})
sunspots["date"] = pd.to_datetime(sunspots[["year", "month", "day"]])
sunspots = sunspots.drop(columns=["year", "month", "day"])
print(sunspots.iloc[10:13])
print(sunspots.dtypes)
print(sunspots["date"].iloc[[0, -1]].tolist())
```

## demo19 · Dates as index

```python
import pandas as pd

filepath = "ISSN_D_tot.csv"
col_names = ["year", "month", "day", "dec_date", "sunspots", "definite"]
sunspots = pd.read_csv(filepath, header=None, names=col_names, na_values={"sunspots": [" -1"]})
sunspots["date"] = pd.to_datetime(sunspots[["year", "month", "day"]])
sunspots = sunspots.drop(columns=["year", "month", "day"])

sunspots = sunspots.set_index("date")
print(sunspots.index.is_unique, sunspots.index.is_monotonic_increasing)
print(sunspots.iloc[10:13])
```

## demo21 · Writing files

```python
import pandas as pd

filepath = "ISSN_D_tot.csv"
col_names = ["year", "month", "day", "dec_date", "sunspots", "definite"]
sunspots = pd.read_csv(filepath, header=None, names=col_names, na_values={"sunspots": [" -1"]})
sunspots["date"] = pd.to_datetime(sunspots[["year", "month", "day"]])
sunspots = sunspots.drop(columns=["year", "month", "day"]).set_index("date")
sunspots = sunspots[["sunspots", "definite"]]

sunspots.to_csv("sunspots.csv")
sunspots.to_csv("sunspots.tsv", sep="\t")
sunspots.to_excel("sunspots.xlsx")

restored = pd.read_csv("sunspots.csv", parse_dates=["date"], index_col="date")
print(restored.dtypes)

summary = pd.DataFrame(
    {
        "statistic": ["mean", "std"],
        "sunspots": [sunspots["sunspots"].mean(), sunspots["sunspots"].std()],
    }
)
summary.to_csv("summary.csv", index=False, float_format="%.2f")
with open("summary.csv") as file:
    print(file.read())
```
