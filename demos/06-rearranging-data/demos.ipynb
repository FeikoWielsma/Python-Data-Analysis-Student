## demo01 · Reshaping by Pivoting

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")
print(trials)

t = trials.pivot(index="treatment", columns="gender", values="response")
print(t)
```

## demo02 · Pivoting Multiple Columns

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")

print(trials)

t = trials.pivot(index="treatment", columns="gender")
print(t)
```

## demo03 · Melt

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")
wide = trials.pivot(index="treatment", columns="gender", values="response")
wide = wide.reset_index()
print(wide)

long = wide.melt(
    id_vars="treatment", value_vars=["F", "M"], var_name="gender", value_name="response"
)
print(long)
```

## demo04 · Creating Multi Level Index

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")

print(trials)

trials = trials.set_index(["treatment", "gender"])
print(trials)
```

## demo05 · Unstacking Multi Index

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")

print(trials)

trials = trials.set_index(["treatment", "gender"])
print(trials)

t = trials.unstack(level="gender")
print(t)

s = trials.unstack(level=1)
print(s)
```

## demo06 · Stacking DataFrames

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv").set_index(["treatment", "gender"])
wide = trials.unstack(level="gender")
restored = wide.stack(level="gender")
print(restored)

sparse = trials.drop(index=("B", "M"))
expanded = sparse.unstack("gender").stack("gender")
print(expanded)
print(len(sparse), len(expanded))
```

## demo07 · Swapping Levels and Sorting Rows

```python
import pandas as pd

trials = pd.read_csv("trials_01.csv")

print(trials)

stacked = trials.set_index(["treatment", "gender"])
print(stacked)

swapped = stacked.swaplevel(0, 1)
print(swapped)

sorted_trials = swapped.sort_index()
print(sorted_trials)
```
