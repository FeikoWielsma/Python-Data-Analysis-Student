## demo01 · Boolean Filter Count and Groupby

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon"],
        "city": ["Austin", "Dallas", "Austin", "Dallas"],
        "bread": [139, 237, 326, 456],
        "butter": [20, 45, 70, 98],
    }
)

print(sales)
print(sales.loc[sales["weekday"] == "Sun"].count())

print(sales.groupby("weekday").count())
```

## demo02 · Groupby and Sum

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon"],
        "city": ["Austin", "Dallas", "Austin", "Dallas"],
        "bread": [139, 237, 326, 456],
        "butter": [20, 45, 70, 98],
    }
)

s = sales.groupby("weekday")["bread"].sum()
print(s)

s = sales.groupby("weekday")[["bread", "butter"]].sum()
print(s)
```

## demo04 · Groupby and Sum by Series

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon"],
        "city": ["Austin", "Dallas", "Austin", "Dallas"],
        "bread": [139, 237, 326, 456],
        "butter": [20, 45, 70, 98],
    }
)
print(sales)
customers = pd.Series(["Dave", "Alice", "Bob", "Alice"])
print(customers)

print(sales.groupby(customers))
print(sales.groupby(customers)["bread"])
print(sales.groupby(customers)["bread"].sum())
```

## demo06 · groupby Multiple Aggregations

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon"],
        "city": ["Austin", "Dallas", "Austin", "Dallas"],
        "bread": [139, 237, 326, 456],
        "butter": [20, 45, 70, 98],
    }
)
print(sales)

print(sales.groupby("city")[["bread", "butter"]].max())

print(sales.groupby("city")[["bread", "butter"]].agg(["max", "sum"]))
```

## demo07 · Custom Aggregation

```python
import pandas as pd

sales = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon"],
        "city": ["Austin", "Dallas", "Austin", "Dallas"],
        "bread": [139, 237, 326, 456],
        "butter": [20, 45, 70, 98],
    }
)
print(sales)

customers = pd.Series(["Dave", "Alice", "Bob", "Alice"])
print(customers)


def data_range(series):
    return series.max() - series.min()


print(sales.groupby("weekday")[["bread", "butter"]].agg(data_range))

print(sales.groupby(customers)[["bread", "butter"]].agg({"bread": "sum", "butter": data_range}))
```

## demo08 · Named Aggregation

```python
import numpy as np
import pandas as pd

# Monday has a missing bread value; the last row has no weekday
sales_na = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon", np.nan],
        "city": ["Austin", "Dallas", "Austin", "Dallas", "Austin"],
        "bread": [139, 237, 326, np.nan, 210],
        "butter": [20, 45, 70, 98, 33],
    }
)

# Each keyword names an output column: name=(column, function)
summary = sales_na.groupby("city").agg(
    rows=("bread", "size"),
    bread_count=("bread", "count"),
    bread_total=("bread", "sum"),
    butter_mean=("butter", "mean"),
)
print(summary)
```

## demo10 · Missing Group Keys

```python
import numpy as np
import pandas as pd

# Monday has a missing bread value; the last row has no weekday
sales_na = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon", np.nan],
        "city": ["Austin", "Dallas", "Austin", "Dallas", "Austin"],
        "bread": [139, 237, 326, np.nan, 210],
        "butter": [20, 45, 70, 98, 33],
    }
)

# Rows with a missing key are dropped by default (dropna=True)
print(sales_na.groupby("weekday").size().sum())
print(len(sales_na))
print(sales_na.groupby("weekday", dropna=False).size())
```

## demo11 · Unobserved Categories

```python
import numpy as np
import pandas as pd

# Monday has a missing bread value; the last row has no weekday
sales_na = pd.DataFrame(
    {
        "weekday": ["Sun", "Sun", "Mon", "Mon", np.nan],
        "city": ["Austin", "Dallas", "Austin", "Dallas", "Austin"],
        "bread": [139, 237, 326, np.nan, 210],
        "butter": [20, 45, 70, 98, 33],
    }
)

# Houston is a known city without sales_na
cities = ["Austin", "Dallas", "Houston"]
sales_na["city"] = pd.Categorical(sales_na["city"], categories=cities)

# pandas 3 defaults to observed=True; pandas 2 defaults to False
print(sales_na.groupby("city", observed=True)["bread"].sum())
print(sales_na.groupby("city", observed=False)["bread"].sum())
print(sales_na.groupby("city", observed=False).size())
```

## demo20 · groupby apply

```python
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")


def chevy_mean(group):
    chevy = group["name"].str.contains("chevrolet")
    return group.loc[chevy, "mpg"].mean()


print(auto.groupby("yr")[["name", "mpg"]].apply(chevy_mean).head(4))


def heaviest(group):
    return group.nlargest(1, "weight")


print(auto.groupby("origin")[["name", "weight"]].apply(heaviest))
```

## demo23 · Z-Score Conventions

```python
import numpy as np
import pandas as pd
from scipy import stats

auto = pd.read_csv("auto-mpg.csv")


def zscore(series):
    return (series - series.mean()) / series.std()


# pandas .std() divides by n - 1 (ddof=1); scipy divides by n (ddof=0)
print(auto["mpg"].std(), auto["mpg"].std(ddof=0))
print(zscore(auto["mpg"]).head(2).to_numpy())
print(stats.zscore(auto["mpg"])[:2])
print(stats.zscore(auto["mpg"], ddof=1)[:2])

# Missing values: pandas skips them; scipy propagates NaN by default
s = pd.Series([18.0, 15.0, np.nan, 16.0])
print(zscore(s).round(3).tolist())
print(stats.zscore(s, ddof=1))
print(stats.zscore(s, ddof=1, nan_policy="omit").round(3))

# One value has no sample standard deviation
print(pd.Series([20.0]).std())

# Grouping by year and cylinders creates eight single-car groups
by_yr_cyl = auto.groupby(["yr", "cyl"])["mpg"]
print((by_yr_cyl.size() == 1).sum())
print(by_yr_cyl.transform(zscore).isna().sum())
```

## demo25 · apply-transformation-and-aggregation

```python
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")
groups = auto.groupby("yr")["mpg"]
mean = groups.transform("mean")
std = groups.transform("std", ddof=1)
result = auto[["yr", "name"]].copy()
result["mpg_z"] = (auto["mpg"] - mean) / std.where(std > 0)
result = result[["mpg_z", "yr", "name"]]
assert result.index.equals(auto.index)
assert result[["yr", "name"]].equals(auto[["yr", "name"]])
assert result["mpg_z"].groupby(result["yr"]).mean().abs().max() < 1e-12
print(result.head())
```

## demo26 · Plot a grouped result

```python
import matplotlib.pyplot as plt
import pandas as pd

auto = pd.read_csv("auto-mpg.csv")
mean_mpg = auto.groupby("yr")["mpg"].mean()

fig, ax = plt.subplots(figsize=(8, 4))
mean_mpg.plot.bar(ax=ax, rot=0)
ax.set(
    title="Average fuel economy rose from 1970 to 1982",
    xlabel="Model year (19xx)",
    ylabel="Miles per gallon",
)
fig.tight_layout()
fig.savefig("mean-mpg-by-year.png", dpi=150)
plt.show()
```
